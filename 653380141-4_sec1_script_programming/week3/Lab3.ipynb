```python
# Multiplication Table Generator

number = int(input("Enter a number: "))

print(f"\nMultiplication Table of {number}\n")

for i in range(1, 13):
    print(f"{number} x {i} = {number * i}")
```

```python
# 12 x 12 Multiplication Grid

print("12 x 12 Multiplication Table\n")

for i in range(1, 13):

    for j in range(1, 13):
        print(f"{i*j:4}", end="")

    print()
```

```python
# Countdown Timer

start = int(input("Enter starting number: "))

while start >= 0:
    print(start)
    start -= 1

print("Blast off!")
```

```python
# Guessing Game

secret_number = 42

while True:

    guess = int(input("Guess the number: "))

    if guess < secret_number:
        print("Too low! Try again.")

    elif guess > secret_number:
        print("Too high! Try again.")

    else:
        print("Congratulations! You guessed it!")
        break
```

```python

# Square Pattern

size = 5

print("Square Pattern\n")

for row in range(size):

    for col in range(size):
        print("*", end=" ")

    print()
```

```python
# Right Triangle Pattern

height = 5

print("Right Triangle\n")

for row in range(1, height + 1):

    for col in range(row):
        print("*", end=" ")

    print()
```

```python
# Inverted Triangle Pattern

height = 5

print("Inverted Triangle\n")

for row in range(height, 0, -1):

    for col in range(row):
        print("*", end=" ")

    print()
```

```python
# Pyramid Pattern

height = 5

print("Pyramid Pattern\n")

for row in range(1, height + 1):

    # พิมพ์ช่องว่าง
    for space in range(height - row):
        print(" ", end="")

    # พิมพ์เครื่องหมาย *
    for star in range(2 * row - 1):
        print("*", end="")

    print()
```
