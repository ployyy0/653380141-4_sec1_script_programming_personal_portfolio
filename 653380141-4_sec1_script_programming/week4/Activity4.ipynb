```python
print("====Activity 4.2====")

todo_list = []

while True:
    print("\n===== TO-DO LIST =====")
    print("1. Add Task")
    print("2. View Tasks")
    print("3. Complete Task")
    print("4. Exit")

    choice = input("Choose: ")

    if choice == "1":
        task = input("Enter task: ")
        todo_list.append(task)
        print("Task added.")

    elif choice == "2":
        if len(todo_list) == 0:
            print("No tasks.")
        else:
            print("\nYour Tasks:")
            for i, task in enumerate(todo_list, start=1):
                print(f"{i}. {task}")

    elif choice == "3":
        if len(todo_list) == 0:
            print("No tasks.")
        else:
            for i, task in enumerate(todo_list, start=1):
                print(f"{i}. {task}")

            index = int(input("Enter task index to complete: "))


            if 1 <= index <= len(todo_list):
                completed = todo_list.pop(index - 1)
                print(f"Completed: {completed}")
            else:
                print("Invalid index.")

    elif choice == "4":
        print("Goodbye!")
        break

    else:
        print("Invalid choice.")
```
