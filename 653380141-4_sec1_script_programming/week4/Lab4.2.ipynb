```python
import math

print("===== Lab 4.2: Geometric Calculations =====")

# ------------------------------------------
# 1. กำหนดพิกัดจุด
# ------------------------------------------
point1 = (3, 4)
point2 = (6, 8)

print("\n--- Access Coordinates ---")
print(f"Point 1: x = {point1[0]}, y = {point1[1]}")
print(f"Point 2: x = {point2[0]}, y = {point2[1]}")

# ------------------------------------------
# 2. ทดลองแก้ไขค่า Tuple (Immutability)
# ------------------------------------------
print("\n--- Tuple Immutability ---")

try:
    point1[0] = 5
except TypeError as error:
    print("Error:", error)
    print("Tuple is immutable, so its values cannot be changed.")

# หลังจากทดลองแล้ว ให้คอมเมนต์บรรทัดนี้ไว้
# point1[0] = 5

# ------------------------------------------
# 3. คำนวณระยะห่าง
# ------------------------------------------
print("\n--- Calculate Distance ---")

distance = math.sqrt(
    (point2[0] - point1[0]) ** 2 +
    (point2[1] - point1[1]) ** 2
)

print(f"Distance between Point1 {point1} and Point2 {point2} = {distance:.2f}")

# ==========================================
# Challenge
# รับค่าพิกัดจากผู้ใช้
# ==========================================

print("\n===== Challenge =====")

x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

point1 = (x1, y1)
point2 = (x2, y2)

distance = math.sqrt(
    (point2[0] - point1[0]) ** 2 +
    (point2[1] - point1[1]) ** 2
)

print("\nResult")
print(f"Point 1 = {point1}")
print(f"Point 2 = {point2}")
print(f"Distance = {distance:.2f}")
```
