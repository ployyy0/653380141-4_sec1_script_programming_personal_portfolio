```python
# ============================================================
# 🔧 Activity 5.2: Word Frequency Counter (using Dictionary)
# ------------------------------------------------------------
# Objective:
# Count the frequency of each word in a sentence using a dictionary.
# ============================================================

# ขั้นตอนที่ 1: รับข้อความจากผู้ใช้
sentence = input("Enter a sentence: ")

# ขั้นตอนที่ 2: แปลงข้อความเป็นตัวพิมพ์เล็ก
sentence = sentence.lower()

# ขั้นตอนที่ 3: แยกข้อความออกเป็นคำ
words = sentence.split()

# ขั้นตอนที่ 4: สร้าง Dictionary ว่าง
word_counts = {}

# ขั้นตอนที่ 5 และ 6: วนลูปนับจำนวนคำ
for word in words:

     # Challenge: ลบเครื่องหมายวรรคตอนก่อนนับคำ
    word = word.strip('.,!?"')

    # ข้ามคำว่าง (ถ้ามี)
    if word == "":
        continue

    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

# ขั้นตอนที่ 7: แสดงผลลัพธ์
print("\n=== Word Frequency ===")
print(word_counts)
```
