Lab 6.1: Simple Calculator Functions

```python
# ฟังก์ชันบวก
def add(a, b):
    """Return the sum of two numbers."""
    return a + b

# ฟังก์ชันลบ
def subtract(a, b):
    """Return the difference between two numbers."""
    return a - b

# ฟังก์ชันคูณ
def multiply(a, b):
    """Return the product of two numbers."""
    return a * b

# ฟังก์ชันหาร
def divide(a, b):
    """Return the division result or an error message."""
    if b == 0:
        return "Error: Division by zero"
    return a / b

# ฟังก์ชันยกกำลัง โดยค่าเริ่มต้นยกกำลัง 2
def power(base, exponent=2):
    """Return base raised to exponent."""
    return base ** exponent
```

```python
# ใช้ while loop เพื่อให้เลือกคำนวณได้หลายครั้ง
while True:
    print("\n===== Simple Calculator =====")
    print("1. Add (+)")
    print("2. Subtract (-)")
    print("3. Multiply (*)")
    print("4. Divide (/)")
    print("5. Power (**)")
    print("0. Exit")

    # รับตัวเลือกจากผู้ใช้
    choice = input("Choose an operation: ")

    # ออกจากโปรแกรมเมื่อเลือก 0
    if choice == "0":
        print("Goodbye!")
        break

    # ตรวจสอบว่าตัวเลือกถูกต้องหรือไม่
    if choice not in ["1", "2", "3", "4", "5"]:
        print("Invalid choice. Please try again.")
        continue

    # รับตัวเลขสองจำนวน
    try:
        num1 = float(input("Enter the first number: "))
        num2 = float(input("Enter the second number: "))
    except ValueError:
        print("Please enter valid numbers.")
        continue

    # เรียกใช้ฟังก์ชันตามตัวเลือก
    if choice == "1":
        result = add(num1, num2)
    elif choice == "2":
        result = subtract(num1, num2)
    elif choice == "3":
        result = multiply(num1, num2)
    elif choice == "4":
        result = divide(num1, num2)
    elif choice == "5":
        result = power(num1, num2)

    # แสดงผลลัพธ์
    print("Result:", result)
```

Lab 6.2: Custom Module and Standard Library

```python
%%writefile my_utils.py

# ฟังก์ชันทักทายผู้ใช้
def greet(name):
    """Print a greeting message."""
    print(f"Hello, {name}! Welcome to my utility functions.")

# ฟังก์ชันคำนวณพื้นที่สี่เหลี่ยมผืนผ้า
def calculate_area_rectangle(length, width):
    """Calculate the area of a rectangle."""
    return length * width

# ฟังก์ชันคำนวณ Factorial
def factorial(n):
    """Calculate the factorial of a non-negative integer."""

    # ตรวจสอบว่าตัวเลขติดลบหรือไม่
    if n < 0:
        return "Factorial is not defined for negative numbers."

    # Factorial ของ 0 มีค่าเท่ากับ 1
    elif n == 0:
        return 1

    else:
        result = 1

        # คูณตัวเลขตั้งแต่ 1 จนถึง n
        for i in range(1, n + 1):
            result *= i

        return result

# ตัวแปรค่าคงที่ในโมดูล
PI_VALUE = 3.14159
```

```python
# นำเข้าโมดูลที่เราสร้างเอง
import my_utils

# นำเข้า Standard Library
import math
import random as rnd

# ทดลองใช้ฟังก์ชันจาก my_utils
my_utils.greet("Students")

# คำนวณพื้นที่สี่เหลี่ยมผืนผ้า
area = my_utils.calculate_area_rectangle(5, 10)
print("Area of rectangle:", area)

# คำนวณ Factorial
num_factorial = my_utils.factorial(5)
print("Factorial of 5:", num_factorial)

# อ่านค่าคงที่จากโมดูล
print("PI value:", my_utils.PI_VALUE)

# ใช้ math หารากที่สอง
print("Square root of 25:", math.sqrt(25))

# แสดงค่าพายจาก math
print("Value of pi:", math.pi)

# สุ่มจำนวนเต็มตั้งแต่ 1 ถึง 100
random_num = rnd.randint(1, 100)
print("Random number:", random_num)

# สุ่มเลือกผลไม้หนึ่งรายการ
fruits = ["apple", "banana", "cherry", "date"]
random_fruit = rnd.choice(fruits)
print("Random fruit:", random_fruit)
```

Activity 6.1: Refactoring a Previous Lab Assignment

```python
# ตรวจสอบว่าเป็นจำนวนบวก ลบ หรือศูนย์
def is_positive_negative_zero(num):
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"

# ตรวจสอบว่าเป็นเลขคู่หรือเลขคี่
def is_even_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"
```

```python
# รับจำนวนเต็มจากผู้ใช้
number = int(input("Enter an integer: "))

# เรียกใช้ฟังก์ชันทั้งสอง
sign_result = is_positive_negative_zero(number)
parity_result = is_even_odd(number)

# แสดงผลลัพธ์
print("Number type:", sign_result)
print("Even or odd:", parity_result)
```
